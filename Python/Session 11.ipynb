```python
def grocery_cost(num_of_apples, num_of_oranges):
    price_of_apples = 3
    cost_of_apples = num_of_apples * price_of_apples
    price_of_oranges = 2
    cost_of_oranges = num_of_oranges * price_of_oranges
    total_cost = cost_of_apples + cost_of_oranges
    return total_cost

gabby=grocery_cost(4,5)
print(gabby)
    
```

```python
def orchard(price_straw, price_cherries, weight_straw, weight_cherries):
    cost_of_straw = price_straw * weight_straw
    cost_of_cherries = price_cherries * weight_cherries
    total_cost2 = cost_of_straw + cost_of_cherries
    return total_cost2
```

```python
# Scenario 1 Terhune Orchard
gabby = orchard(7,10,4,3)
naeem = orchard(7,10,7,4)
omar = orchard(7,10,5,3)
chris= gabby + omar + naeem
print(chris)
```

```python
# Scenario 2: Mamoun's falafel
'''
falafel = $3 per bag
Lentil soup = $2 per bowl
Pita = $1 per piece

function = lunch
return = cost_lunch
variables = price_falafel, amount_falafel, price_soup, amount_soup, price_pita, amount_pita

Scenario : 
Darius : 2F, 2S, 1P
Belle : 1F, 1S, 4P
Amel : 3F, 2S, 3P
'''
# takes in the price and amounts of food and returns the total cost
def lunch(price_falafel, amount_falafel, price_soup, amount_soup, price_pita, amount_pita):
    
    cost_of_falafel = price_falafel * amount_falafel
    cost_of_soup = price_soup * amount_soup
    cost_of_pita = price_pita * amount_pita

    # Special deal: cost of falafel is halved (50% off)
    cost_falafel_deal = price_falafel * 0.5 * amount_falafel
    # Or we could have said: cost_falafel_deal = cost_falafel/2
    total_lunch_cost = cost_of_falafel + cost_of_soup + cost_of_pita
    total_lunch_cost_deal = cost_falafel_deal + cost_of_soup + cost_of_pita
    return total_lunch_cost

# Lunch Scenario
darius = lunch(3,2,2,2,1,1)
belle = lunch(3,1,2,1,1,4)
amel = lunch(3,3,2,2,1,3)

# How much does Chris has to pay?
chris = darius + belle + amel
print(chris)
```

```python
'''
F = C x (9/5) +32
function = cels_to_fahr
return = fahrenheit
variables = celsius
'''
def cels_to_fahr(temp_in_cels):
    temp_in_fahr = temp_in_cels * (9/5) + 32
    return temp_in_fahr

celsius_temp = 15
converted_temp = cels_to_fahr(celsius_temp)
print(converted_temp)
```
