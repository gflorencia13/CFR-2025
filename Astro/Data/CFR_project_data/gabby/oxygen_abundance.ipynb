```python
# Import the libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from scipy import stats

# Read in the data
oxy_hydro = pd.read_csv('../gabby/oxygen_hydrogen_2.9.csv')
```

```python
oxy_hydro
```

**Total of rows and columns**  
**Rows**: 5816056   
**Columns**:  2

```python
mass_oxy = oxy_hydro['Mass Oxygen / Star mass']
mass_hydro = oxy_hydro['Mass Hydrogen / Star mass']
```

```python
# Divide the oxygen by the hydrogen
divide_oxy_hydro = mass_oxy / mass_hydro
divide_oxy_hydro
```

```python
# Multiply by 1.008 and divide by 15.999
multiply_divide = divide_oxy_hydro * 1.008 / 15.999
multiply_divide
```

```python
# Take log10 of the result
log_10 = np.log10(multiply_divide)
log_10
```

```python
# Add 12 to it
add_12 = 12 + log_10
add_12 # 12 + log[O/H]
```

```python
# Set up first before plotting
import matplotlib.colors as mcolors

plt.figure(figsize=(12,8), facecolor='black')
plt.gca().set_facecolor('black')
sns.set_context("poster")

# Make the histogram
n, bins, patches = plt.hist(add_12, bins=50, color='cyan', edgecolor = 'white', linewidth=0.5, alpha = 0.9)
plt.title('Distribution of Oxygen Abundances in Simulated Galaxies', color='white', fontweight='bold', pad=20)
plt.xlabel('Oxygen Abundance (12 + log [O/H])', color='white', fontweight='bold')
plt.ylabel('Number of Galaxies', color='white', fontweight='bold')
plt.xlim(4,10)

# Define your custom colors (example: blue to red gradient)
custom_colors = ['#b7b2ed', '#6f65e0', '#2b1dcc']  # Replace with your desired hex colors
custom_cmap = mcolors.LinearSegmentedColormap.from_list('custom', custom_colors)

# Normalize bins
norm = plt.Normalize(n.min(), n.max()) 

# Apply custom colormap
for patch in patches:
    color = custom_cmap(norm(patch.get_height()))
    patch.set_facecolor(color)

# Make everything white for dark theme
plt.xlim(5,10)
plt.tick_params(colors='white')  # White tick marks
plt.legend(facecolor='black', edgecolor='white', labelcolor='white')

# Clean up the borders
ax = plt.gca()
ax.spines['bottom'].set_color('white')
ax.spines['left'].set_color('white')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Cecilia galaxy line
plt.axvline(8.05, label='Cecilia Galaxy', linewidth=4, color='crimson', linestyle='--', alpha=0.9)

# Create bell curve that matches histogram scale
x = np.linspace(4, 10, 100)
mean = np.mean(add_12)
std = np.std(add_12)


#plt.grid()
plt.legend()

plt.savefig('Presentation plot.png')
plt.show()
#plt.tight_layout()
```
