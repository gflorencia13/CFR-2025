## Part 1: Basic Plotting with Pandas (30 minutes)

Pandas makes it easy to create basic plots using the `plot` method. The default plot type is a line plot.

#### Example 1: Line Plot


```python
# Import dataset
df = pd.read_csv ("data_tech_order.csv")

# Line plot
df.plot(x='Weight', y='Price', kind='line')
plt.title("Price vs Weight")
plt.xlabel("Weight (pounds)")
plt.ylabel("Price ($)")
```

#### Example 2: Customizing Line Plot

You can customize your plots by changing colors, adding markers, and more.


```python
# Import dataset
df = pd.read_csv ("data_tech_order.csv")

# Line plot
df.plot(x='Weight', y='Price', kind='line', color="yellow", marker="s", linestyle="--")
plt.title("Price vs Weight")
plt.xlabel("Weight (pounds)")
plt.ylabel("Price ($)")
```

#### Example 4: Histogram

Histograms are useful for displaying the distribution of a dataset.


```python
help(df.hist)
```

```python
# Generate our data
# **************************
# STEP 1: generate 100 random numbers
data = np.random.randn(1000) # array
data[:11]

# STEP 2: create dataframe with the data
df = pd.DataFrame(data, columns=["Values"])
df.head(10)

# STEP 3: save the DataFrame to a csv file (can be used in Excel)
# Without variable option
# Create a variable for csv file name so that I only need to change the name in one place
name_csv_file = "data_session10_example4.csv"
df.to_csv(name_csv_file, index= False) # false is to add another number 

# STEP 4: Bring in the data from the csv
df_from_csv = pd.read_csv(name_csv_file)
df_from_csv.head(10)

#Create plot
# ********************************
plt.hist(df_from_csv["Values"], bins=30, edgecolor='paleturquoise',color="lightcyan")
plt.title("Histogram of Values")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.grid()
```

## Part 3: Advanced Customization and Subplots (30 minutes)


#### Example 6: Creating Subplots

Subplots allow you to create multiple plots in a single figure.

```python
# Sample Data
x = [1, 2, 3, 4, 5]
y1 = [2, 3, 5, 7, 11]
y2 = [1, 4, 6, 8, 10]

# Create a DataFrame from Sample Data
df = pd.DataFrame({'X':x, 'Y1':y1, 'Y2':y2,})
print(df)

# Create graphs
# row, column, plot number
plt.subplot(3,2,1)
plt.plot(df['X'],df ['Y1'], color='r')

plt.subplot(3,2,2)
plt.plot(df['X'],df ['Y1'], color='orange')

plt.subplot(3,2,3)
plt.plot(df['X'],df ['Y1'], color='g')

plt.subplot(3,2,4)
plt.plot(df['X'],df ['Y1'], color='b')

plt.subplot(3,2,5)
plt.plot(df['X'],df ['Y1'], color='m')

plt.subplot(3,2,6)
plt.plot(df['X'],df ['Y1'], color='purple')
```

**Practice Problem 6: Create Your Own Subplots**

Create a figure with two subplots:
- A line plot for the data x: [0, 1, 2, 3, 4], y: [0, 1, 4, 9, 16]
- A bar plot for the data categories: ['A', 'B', 'C'], values: [5, 10, 15]

**Solution:**


```python
x = [0, 1, 2, 3, 4]
y1 = [0, 1, 4, 9, 16]
x2 = ['A', 'B', 'C']
y2 = [5, 10, 15]

# Create a DataFrame from Sample Data
df1 = pd.DataFrame({'X':x, 'Y1':y1})
df2 = pd.DataFrame({'X2':x2, 'Y2':y2})
print(df1)
print(df2)

# Create graphs
# row, column, plot number
plt.subplot(2,4,1)
plt.plot(df1['X'], df1['Y1'], color='r')

plt.subplot(2,4,2)
plt.bar(df2['X2'], df2['Y2'], color='lime')

plt.subplot(2,4,3)
plt.plot(df1['X'], df1['Y1'], color='m')

plt.subplot(2,4,4)
plt.bar(df2['X2'], df2['Y2'], color='royalblue')

plt.subplot(2,4,5)
plt.plot(df1['X'], df1['Y1'], color='pink')

plt.subplot(2,4,6)
plt.bar(df2['X2'], df2['Y2'], color='slategrey')

plt.subplot(2,4,7)
plt.plot(df1['X'], df1['Y1'], color='cyan')

plt.subplot(2,4,8)
plt.bar(df2['X2'], df2['Y2'], color='wheat')
```
