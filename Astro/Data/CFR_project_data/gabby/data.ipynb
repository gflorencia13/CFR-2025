```python
# List all of the arrays
x = [0, 0.5, 1, 1.5, 2, 2.4, 2.9]
y = [mean_data0, mean_data05, mean_data1, mean_data15, mean_data2, mean_data24, mean_data29]
error = [std_data0 / np.sqrt(len(mass0)), std_data05 / np.sqrt(len(mass05)), std_data1 / np.sqrt(len(mass1)), std_data15 / 
         np.sqrt(len(mass15)) , std_data2 /  np.sqrt(len(mass2)), std_data24 /  np.sqrt(len(mass24)), std_data29 / np.sqrt(len(mass29)) ]     
```

```python
# Create the plot
plt.scatter(x,y, color='slateblue')
plt.title('Mass of Galaxies vs Time Ago')
plt.xlabel('Billions of Years Ago')
plt.ylabel('Average Mass of Galaxies')

# Error bar
plt.errorbar(x,y, yerr=error, fmt='o', capsize=5, elinewidth=1, capthick=2 )
plt.grid()
```
