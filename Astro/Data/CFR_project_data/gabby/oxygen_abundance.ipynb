```python
# Import the libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Read in the data
oxy_hydro = pd.read_csv('../gabby/oxygen_hydrogen_2.9.csv')
```

```python
oxy_hydro
```

**Total of rows and columns**  
**Rows**: 5816056   
**Columns**:  2

```python
mass_oxy = oxy_hydro['Mass Oxygen / Star mass']
mass_hydro = oxy_hydro['Mass Hydrogen / Star mass']
```

```python
# Divide the oxygen by the hydrogen
divide_oxy_hydro = mass_oxy / mass_hydro
divide_oxy_hydro
```

```python
# Multiply by 1.008 and divide by 15.999
multiply_divide = divide_oxy_hydro * 1.008 / 15.999
multiply_divide
```

```python
# Take log10 of the result
log_10 = np.log10(multiply_divide)
log_10
```

```python
# Add 12 to it
add_12 = 12 + log_10
add_12 # 12 + log[O/H]
```

```python
plt.hist(add_12, bins=200, color='teal', alpha = 0.6)
plt.title('Distribution of Oxygen Abundances in Simulated Galaxies')
plt.xlabel('Oxygen Abundance (12 + log [O/H])')
plt.ylabel('Number of Galaxies')
plt.xlim(4,10)
plt.grid()

plt.axvline(8.05, label='Cecilia Galaxy')

plt.grid()
plt.legend()
```
