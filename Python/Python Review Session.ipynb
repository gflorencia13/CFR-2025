## If Statements

If statements and conditionals let your code make decisions.

If a condition is true, run this part of code.

If it's false, run a different part of code.

```python
if condition:
    # code if this condition is true and will run
elif another_condition:
    # code if first condition is false, and this other condition is true will run
else:
    # code if none of the above are true
```

```python
if condition:
    if another_condition:
        # Code will run if BOTH if statements are true.
else:
    # Code will run if first if statement is not true
```

Example with numbers

```python
temperature = 67

if temperature > 85:
    print("It's hot!")
elif temperature >= 60:
    print("It's nice.")
else:
    print("It's cold.")
```

## While Loops

While loops tell the computer to run a piece of code as long as a certain condition is true.  
Think of it like: "Keep doing this code while this is true"

count = 1
while count <= 5:
    print("Count is", count)
    count = count + 1 
# Updating the variable is very important so we don't run into infinite loops

```python
password = ""

while password != "Hello":
    password = input("Enter the password: ")
    
print("Access granted.")
```

## For Loops

Similar to while loop in the sense that it repeats code.  
A for loop goes through items in a list of range, and runs code for each item.

```python
fruits = ["apple", "orange", "banana", "cherry"]

# For each specific fruit in fruits
for fruit in fruits:
    print("I like" ,fruit)
```

Second way : Using range function.  
range(n) gives you a list of numbers from 0 to n-1.

```python
for x in range(5):
    print(x)
```

Third way : Using range function but choosing where to start and stop.  
range(start, stop)

```python
for x in range(4):
    print(x)
```

Fourth way : range(start, stop, step)  
step is increase by ...

```python
for x in range(2,11,2):
    print(x)
```

Function with a parameter (input)

```python
# Create a function
def greet(name):
    greeting = "Hi "+ name + "! Welcome."
    return greeting
```

```python
# Call the function
print(greet("Darius"))
print(greet("Naeem"))
```

```python
# Create a function
def is_even(number):
    if number % 2 == 0: # mod 2 means the remainder when dividing by 2
        return True
    else:
        return False
```

```python
# Using the function

print(is_even(36))

is_33_even = is_even(33)
print(is_33_even)
```

```python
def total_flowers_cost(amount_of_flowers, price_of_flowers):
    total_cost = amount_of_flowers * price_of_flowers
    return total_cost
```

```python
Amel = total_flowers_cost(3,12)
print("Amel spends", Amel, " on flowers.")

Belle = total_flowers_cost(4,10)
print("Belle spends", Belle, "on flowers.")
```

## Matplotlib

Libraries = a collection of a bunch of functions, you need to import a library if you want to use it's function.  

Matplotlib is a specific library that people use to make plots and graphs.

We have been using pyplot section of Matplotlib.

```python
import matplotlib.pyplot as plt
```

```python
list_1 = [1, 2, 3, 4]
list_2 = [10, 20, 30, 40]

plt.plot(list_1, list_2, color="green", marker='o', markersize=8, linewidth=2) # create a plot
plt.title("My first plot")
plt.xlabel("X values")
plt.ylabel("Y values")
```

Bar Plots:  
Used for when you have one categorical data set and one numerical data set.

```python
# Sample data
categories = ['Apples', 'Bananas', 'Cherries']
values = [10, 24, 18]

plt.bar(categories, values)
plt.title("Fruit Sales")
plt.xlabel("Fruit Type")
plt.ylabel("Number Sold")
```

```python
# Sample data
categories = ['Apples', 'Bananas', 'Cherries']
values = [10, 24, 18]
# for different colors, like in array
colors = ['lightblue', 'pink', 'beige']

plt.bar(categories, values, color= colors) #all the same color
plt.title("Fruit Sales")
plt.xlabel("Fruit Type")
plt.ylabel("Number Sold")
```

Histograms look similar to bar plots.  
However, they are used to group NUMERIC data into bins (ranges) and show how many fall into that range.  
Only need ONE data

```python
# Example data of scores
scores = [87, 85, 90, 92, 77, 100, 78, 50, 60, 79, 100]

plt.hist(scores, bins=8, color='skyblue', edgecolor='black')
plt.title("Distribution of Test Scores")
plt.xlabel("Score Ranges")
plt.ylabel("Number of Students")
```

Scatter Plots:  
Similar to line plot, shows the relationship between two sets of numbers.  
Each point represents one data pair.

```python
#Sample data
hours = [1, 2, 3, 4, 5, 6]
scores = [50, 67, 89, 90, 68, 88]

plt.scatter(hours, scores, color='purple')
plt.title("Study Time vs Test Score")
plt.xlabel("Hours of Study")
plt.ylabel("Scores")
plt.grid(True)
```

Subplots in Matplotlib:  
Show multiple plots in one figure

```python
#Sample data

# 1st subplot data
x_subplot1 = [1, 2, 3, 4]
y_subplot1 = [2, 4, 6, 8]

# 2nd subplot data
x_subplot2 = ['Apples', 'Bananas', 'Cherries']
y_subplot2 = [6, 8, 10]

plt.subplot(1, 2, 1) # rows, columns, which number plot you want to create
plt.plot(x_subplot1, y_subplot1, color= 'pink')
plt.annotate ("Peak", xy= (4, 8), xytext = (3, 8), arrowprops=dict(facecolor='skyblue')) 
plt.title("First Subplot")
plt.xlabel("X data")
plt.ylabel("Y data")

plt.subplot(1, 2, 2) # 1 row, 2 columns and now I want to make the second plot
plt.bar(x_subplot2, y_subplot2, color= 'purple')
plt.title("Fruits Sold")
plt.xlabel("Fruits")
plt.ylabel("Number Sold")

plt.tight_layout() # fixes the layout
```

Filtering Data

```python
filtered_data = df[df['Price'] > 600]
print(filtered_data)
```

```python
filtered_data_2 = df[(df['Price'] > 600) & (df['Quantity'] < 80)] # both conditions must be true
print(filtered_data_2)
```

```python
filtered_data_3= df[(df['Price'] > 600) | (df['Quantity'] < 80)] # | is or
print(filtered_data_3)
```

```python
df['Discount'] = [10, 15, 20] # add a column
# If you want to delete the column 
# inplace= True changes the dataframe that we have already
df.drop('Discount', axis = 1, inplace= True) # axis = 1 remove column

df.drop((0), axis=0, inplace=True) # remove a row with the drop function
print(df)
```
